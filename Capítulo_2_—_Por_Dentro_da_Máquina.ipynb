```python
nome = "Andre Ferreira"

print("MEU NOME EM BITS")
print("----------------")

for caractere in nome:
    decimal = ord(caractere)
    binario = format(decimal, "08b")

    print(
        "Caractere:",
        repr(caractere),
        "| Decimal:",
        decimal,
        "| Binário:",
        binario
    )
```

```python

nome = "Andre Ferreira"

print("Conversão do meu nome para representação digital")
print()

for caractere in nome:
    decimal = ord(caractere)
    binario = format(decimal, "08b")

    print(
        repr(caractere),
        "→ Decimal:",
        decimal,
        "→ Binário:",
        binario
    )

print()
print("Conclusão:")
print("Cada caractere pode ser representado por um código numérico.")
print("Esse número também pode ser representado em forma binária usando 0 e 1.")
```
